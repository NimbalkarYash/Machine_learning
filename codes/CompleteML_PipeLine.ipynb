```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("car-sales-extended-missing-data.csv")
```

```python
df.head()
```

```python
df.isna().sum()
```

```python
#getting data ready

from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder
```

```python
#loading random forest regressor

from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import train_test_split
```

```python
#setting up random seed using numpy
np.random.seed(42)
```

```python
#setting up the dataset by deleting the rows with missing price values

df.dropna(subset = ['Price'], inplace = True)
X = df.drop("Price", axis = 1)
Y = df["Price"]
```

```python
df.isnull().sum()
```

```python
#defining categorical data

categorical_features = ["Make", "Colour"]
categorical_transform = Pipeline(
    steps = [
        ('imputer', SimpleImputer(strategy = "constant", fill_value = "missing")),
        ('onehot', OneHotEncoder(handle_unknown = "ignore"))          
])

#definig door data because most of them have 4 doors
door_features = ["Doors"]
door_transform = Pipeline(
    steps = [
        ('imputer', SimpleImputer(strategy = "constant",fill_value = 4 )),
        
    ]
)

#Numeric data 
numeric_features = ["Odometer (KM)"]
numeric_transform = Pipeline(
    steps = [
        ("imputer", SimpleImputer(strategy = "mean"))
    ]
)
```

```python
#preprocessing of data

preprocessor = ColumnTransformer(
    transformers =  [
        ("cat", categorical_transform, categorical_features),
        ("door", door_transform, door_features),
        ("numeric", numeric_transform, numeric_features)
    ]
)
```

```python
df.head()
```

```python
#Creating a preprocessing and modeling pipeline

ModelPipeline = Pipeline(
    steps = [
        ("pre-processor", preprocessor),
        ("Model", RandomForestRegressor())
    ]
)
```

```python
#creating the test train split

X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size = 0.3)
```

```python
ModelPipeline.fit(X_train,Y_train)
ModelPipeline.score(X_test, Y_test)
```
